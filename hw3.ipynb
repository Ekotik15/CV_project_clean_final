```python
import os
import numpy as np
import matplotlib.pyplot as plt
from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras.preprocessing.image import load_img, img_to_array, ImageDataGenerator
from sklearn.metrics import confusion_matrix, classification_report, roc_curve
import seaborn as sns

# ========== 1. ЗАГРУЗКА ДАННЫХ ==========
def load_images_from_folder(folder, target_size=(128, 128)):
    images = []
    filenames = []
    for filename in sorted(os.listdir(folder)):
        if filename.endswith(('.jpg', '.png', '.jpeg')):
            img_path = os.path.join(folder, filename)
            img = load_img(img_path, target_size=target_size, color_mode='rgb')
            img_array = img_to_array(img) / 255.0
            images.append(img_array)
            filenames.append(filename)
    return np.array(images), filenames

# Загрузка тренировочных данных (без проливов)
train_images, _ = load_images_from_folder('dataset/train')
print(f"Загружено тренировочных изображений: {train_images.shape}")

# Загрузка изображений с проливами для определения порога
proliv_images, _ = load_images_from_folder('dataset/proliv')
print(f"Загружено изображений с проливами: {proliv_images.shape}")

# Загрузка тестовых данных
test_images, test_filenames = load_images_from_folder('dataset/test/imgs')
print(f"Загружено тестовых изображений: {test_images.shape}")

# Загрузка разметки тестовых данных
def load_test_labels(annotation_path):
    labels = {}
    with open(annotation_path, 'r') as f:
        for line in f:
            line = line.strip()
            if line:
                # Пробуем разные разделители
                if '\t' in line:
                    parts = line.split('\t')
                elif ' ' in line:
                    parts = line.split()
                elif ',' in line:
                    parts = line.split(',')
                else:
                    continue
                
                if len(parts) >= 2:
                    filename = parts[0]
                    label = parts[1]
                    try:
                        labels[filename] = int(label)
                    except ValueError:
                        # Если метка не число, пробуем преобразовать строку
                        if label.lower() in ['proliv', 'пролив', '1', 'true']:
                            labels[filename] = 1
                        else:
                            labels[filename] = 0
    return labels

test_labels_dict = load_test_labels('dataset/test/test_annotation.txt')
print(f"Загружено {len(test_labels_dict)} меток из аннотации")

# Простые варианты сопоставления - пробуем разные форматы имен
test_labels = []
valid_indices = []

for i, filename in enumerate(test_filenames):
    # Варианты имен для поиска в аннотации
    base_name = os.path.splitext(filename)[0]  # убираем расширение
    
    # Варианты для поиска
    possible_keys = [
        base_name,                           # "test_000"
        base_name.replace('test_', ''),      # "000"
        base_name.replace('test_', '') + '.jpg',  # "000.jpg"
        filename,                            # "test_000.jpg"
        str(i),                              # "0"
        str(i).zfill(3),                     # "000"
        f"{i:03d}",                          # "000"
        f"{i:03d}.jpg",                      # "000.jpg"
    ]
    
    for key in possible_keys:
        if key in test_labels_dict:
            test_labels.append(test_labels_dict[key])
            valid_indices.append(i)
            break
    else:
        # Если не нашли, можно пропустить или использовать дефолтное значение
        print(f"Внимание: не найдена метка для файла {filename}")

if len(test_labels) > 0:
    test_labels = np.array(test_labels)
    test_images = test_images[valid_indices]
    test_filenames = [test_filenames[i] for i in valid_indices]
    print(f"Найдено меток для {len(test_labels)} изображений")
else:
    print("Не найдено ни одной метки. Будем работать без них.")
    test_labels = np.zeros(len(test_images))  # создаем фиктивные метки

# ========== 2. ПОСТРОЕНИЕ И ОБУЧЕНИЕ АВТОЭНКОДЕРА ==========
input_shape = (128, 128, 3)

# Улучшенная модель с большей глубиной
def build_deeper_autoencoder(input_shape=(128, 128, 3)):
    encoder_input = keras.Input(shape=input_shape, name='img')
    
    # Энкодер
    x = layers.Conv2D(32, (3, 3), activation='relu', padding='same')(encoder_input)
    x = layers.Conv2D(32, (3, 3), activation='relu', padding='same')(x)
    x = layers.MaxPooling2D((2, 2), padding='same')(x)
    
    x = layers.Conv2D(64, (3, 3), activation='relu', padding='same')(x)
    x = layers.Conv2D(64, (3, 3), activation='relu', padding='same')(x)
    x = layers.MaxPooling2D((2, 2), padding='same')(x)
    
    x = layers.Conv2D(128, (3, 3), activation='relu', padding='same')(x)
    x = layers.Conv2D(128, (3, 3), activation='relu', padding='same')(x)
    encoded = layers.MaxPooling2D((2, 2), padding='same')(x)
    
    # Декодер
    x = layers.Conv2D(128, (3, 3), activation='relu', padding='same')(encoded)
    x = layers.Conv2D(128, (3, 3), activation='relu', padding='same')(x)
    x = layers.UpSampling2D((2, 2))(x)
    
    x = layers.Conv2D(64, (3, 3), activation='relu', padding='same')(x)
    x = layers.Conv2D(64, (3, 3), activation='relu', padding='same')(x)
    x = layers.UpSampling2D((2, 2))(x)
    
    x = layers.Conv2D(32, (3, 3), activation='relu', padding='same')(x)
    x = layers.Conv2D(32, (3, 3), activation='relu', padding='same')(x)
    x = layers.UpSampling2D((2, 2))(x)
    
    decoded = layers.Conv2D(3, (3, 3), activation='sigmoid', padding='same')(x)
    
    autoencoder = keras.Model(encoder_input, decoded, name='deeper_autoencoder')
    autoencoder.compile(optimizer='adam', loss='mse')
    
    return autoencoder

# Создаем и обучаем модель
print("\nСоздание и обучение автоэнкодера...")
autoencoder = build_deeper_autoencoder(input_shape)
autoencoder.summary()

# Аугментация данных для увеличения тренировочной выборки
datagen = ImageDataGenerator(
    rotation_range=10,
    width_shift_range=0.1,
    height_shift_range=0.1,
    zoom_range=0.1,
    horizontal_flip=True,
    fill_mode='nearest'
)

# Подготовка данных для аугментации
print("Аугментация данных...")
augmented_images = []
for img in train_images:
    img_expanded = np.expand_dims(img, axis=0)
    aug_iter = datagen.flow(img_expanded, batch_size=1)
    for _ in range(2):  # создаем 2 аугментированных версии каждого изображения
        batch = next(aug_iter)  # Исправлено: next(iterator) вместо iterator.next()
        augmented = batch[0]
        augmented_images.append(augmented)

augmented_images = np.array(augmented_images)
all_train_images = np.concatenate([train_images, augmented_images], axis=0)
print(f"Обучающих изображений после аугментации: {all_train_images.shape[0]}")

# Обучение модели
print("\nНачинаем обучение модели...")
history = autoencoder.fit(
    all_train_images, all_train_images,
    epochs=30,
    batch_size=16,
    validation_split=0.1,
    shuffle=True,
    verbose=1
)

# Визуализация обучения
plt.figure(figsize=(10, 5))
plt.plot(history.history['loss'], label='train_loss')
plt.plot(history.history['val_loss'], label='val_loss')
plt.xlabel('Epoch')
plt.ylabel('Loss (MSE)')
plt.legend()
plt.title('Обучение автоэнкодера')
plt.grid(True)
plt.show()

# ========== 3. ОПРЕДЕЛЕНИЕ ОПТИМАЛЬНОГО ПОРОГА ==========
def calculate_mse(autoencoder, images):
    reconstructions = autoencoder.predict(images, verbose=0)
    mse = np.mean(np.square(images - reconstructions), axis=(1, 2, 3))
    return mse

# Вычисление MSE
print("\nВычисление MSE для определения порога...")
train_mse = calculate_mse(autoencoder, train_images)
proliv_mse = calculate_mse(autoencoder, proliv_images)

print(f"MSE на тренировочных данных: среднее={np.mean(train_mse):.6f}, std={np.std(train_mse):.6f}")
print(f"MSE на проливах: среднее={np.mean(proliv_mse):.6f}, std={np.std(proliv_mse):.6f}")

# Метод 1: Стандартный порог (среднее + 2*std)
threshold_2sigma = np.mean(train_mse) + 2 * np.std(train_mse)

# Метод 2: Строгий порог (среднее + 3*std)
threshold_3sigma = np.mean(train_mse) + 3 * np.std(train_mse)

# Метод 3: Оптимальный порог по ROC-кривой
all_images = np.concatenate([train_images, proliv_images], axis=0)
all_labels = np.concatenate([np.zeros(len(train_images)), np.ones(len(proliv_images))], axis=0)
all_mse = calculate_mse(autoencoder, all_images)

fpr, tpr, thresholds = roc_curve(all_labels, all_mse)
optimal_idx = np.argmax(tpr - fpr)
threshold_optimal = thresholds[optimal_idx]

# Метод 4: Порог на основе квантилей
threshold_quantile = np.percentile(train_mse, 95)

print("\nРазличные пороги классификации:")
print(f"1. Среднее + 2σ: {threshold_2sigma:.6f}")
print(f"2. Среднее + 3σ: {threshold_3sigma:.6f}")
print(f"3. Оптимальный (ROC): {threshold_optimal:.6f}")
print(f"4. 95-й квантиль: {threshold_quantile:.6f}")

# Визуализация распределения MSE
plt.figure(figsize=(12, 6))
plt.hist(train_mse, bins=30, alpha=0.7, label='Без пролива (train)', density=True, color='blue')
plt.hist(proliv_mse, bins=30, alpha=0.7, label='С проливом (proliv)', density=True, color='red')

# Рисуем все пороги
thresholds_list = [
    (threshold_2sigma, 'Среднее + 2σ', 'green', '--'),
    (threshold_3sigma, 'Среднее + 3σ', 'orange', '-.'),
    (threshold_optimal, 'Оптимальный (ROC)', 'purple', ':'),
    (threshold_quantile, '95-й квантиль', 'brown', '--')
]

for threshold, label, color, linestyle in thresholds_list:
    plt.axvline(threshold, color=color, linestyle=linestyle, linewidth=2, label=f'{label} = {threshold:.4f}')

plt.xlabel('MSE')
plt.ylabel('Плотность')
plt.legend()
plt.title('Распределение MSE и различные пороги классификации')
plt.grid(True, alpha=0.3)
plt.show()

# ========== 4. ТЕСТИРОВАНИЕ С РАЗНЫМИ ПОРОГАМИ ==========
test_mse = calculate_mse(autoencoder, test_images)

print("\n=== СРАВНЕНИЕ РАЗНЫХ ПОРОГОВ НА ТЕСТЕ ===")

thresholds_to_test = {
    'Среднее + 2σ': threshold_2sigma,
    'Среднее + 3σ': threshold_3sigma,
    'Оптимальный (ROC)': threshold_optimal,
    '95-й квантиль': threshold_quantile
}

best_tnr = 0
best_threshold_name = ''
best_predictions = None

for name, thresh in thresholds_to_test.items():
    predictions = (test_mse > thresh).astype(int)
    
    if len(test_labels) > 0 and len(np.unique(test_labels)) > 1:
        cm = confusion_matrix(test_labels, predictions)
        
        if cm.size == 4:
            tn, fp, fn, tp = cm.ravel()
            tpr = tp / (tp + fn) if (tp + fn) > 0 else 0
            tnr = tn / (tn + fp) if (tn + fp) > 0 else 0
            
            print(f"\n{name} (порог={thresh:.6f}):")
            print(f"  TPR (Recall): {tpr:.4f}")
            print(f"  TNR: {tnr:.4f}")
            print(f"  Accuracy: {(tp+tn)/(tp+tn+fp+fn):.4f}")
            print(f"  FP: {fp}, FN: {fn}")
            
            # Запоминаем лучший порог по TNR
            if tnr > best_tnr:
                best_tnr = tnr
                best_threshold_name = name
                best_predictions = predictions
    else:
        print(f"\n{name} (порог={thresh:.6f}):")
        print(f"  Прогнозы: {sum(predictions)} проливов, {len(predictions)-sum(predictions)} без пролива")

# Используем лучший порог
if best_threshold_name:
    print(f"\nЛучший порог по TNR: {best_threshold_name} (TNR={best_tnr:.4f})")
    final_predictions = best_predictions
    final_threshold = thresholds_to_test[best_threshold_name]
else:
    # Если нет меток, используем оптимальный порог по ROC
    final_predictions = (test_mse > threshold_optimal).astype(int)
    final_threshold = threshold_optimal
    print(f"\nИспользуем оптимальный порог по ROC: {final_threshold:.6f}")

# ========== 5. ВИЗУАЛИЗАЦИЯ РЕЗУЛЬТАТОВ ==========
def visualize_results(images, mse_values, labels, predictions, num_samples=5):
    if len(images) == 0:
        print("Нет изображений для визуализации")
        return
        
    if num_samples > len(images):
        num_samples = len(images)
    
    # Выбираем примеры разных типов
    normal_indices = np.where(predictions == 0)[0]
    anomaly_indices = np.where(predictions == 1)[0]
    
    # Берем по половине из каждого класса
    half_samples = num_samples // 2
    indices = []
    
    if len(normal_indices) > 0:
        indices.extend(np.random.choice(normal_indices, min(half_samples, len(normal_indices)), replace=False))
    if len(anomaly_indices) > 0:
        indices.extend(np.random.choice(anomaly_indices, min(half_samples, len(anomaly_indices)), replace=False))
    
    # Если не набрали достаточно, добавляем случайные
    if len(indices) < num_samples:
        remaining = num_samples - len(indices)
        remaining_indices = np.setdiff1d(np.arange(len(images)), indices)
        if len(remaining_indices) > 0:
            indices.extend(np.random.choice(remaining_indices, min(remaining, len(remaining_indices)), replace=False))
    
    if len(indices) == 1:
        fig, axes = plt.subplots(1, 3, figsize=(15, 4))
        axes = [axes]
    else:
        fig, axes = plt.subplots(len(indices), 3, figsize=(15, 4*len(indices)))
    
    for i, idx in enumerate(indices):
        if len(indices) == 1:
            ax0, ax1, ax2 = axes
        else:
            ax0, ax1, ax2 = axes[i]
        
        # Оригинальное изображение
        ax0.imshow(images[idx])
        truth_label = "Пролив" if labels[idx] == 1 else "Нет пролива" if len(labels) > 0 and len(np.unique(labels)) > 1 else "Неизвестно"
        ax0.set_title(f'Оригинал\nФакт: {truth_label}')
        ax0.axis('off')
        
        # Реконструкция
        reconstruction = autoencoder.predict(images[idx:idx+1], verbose=0)
        ax1.imshow(reconstruction[0])
        ax1.set_title(f'Реконструкция\nMSE: {mse_values[idx]:.6f}')
        ax1.axis('off')
        
        # Разность
        diff = np.abs(images[idx] - reconstruction[0])
        ax2.imshow(diff)
        pred_label = "Пролив" if predictions[idx] == 1 else "Нет пролива"
        ax2.set_title(f'Разность\nПредсказание: {pred_label}')
        ax2.axis('off')
    
    plt.tight_layout()
    plt.show()

# Визуализация результатов
if len(test_images) > 0:
    print("\nВизуализация результатов классификации...")
    visualize_results(test_images, test_mse, 
                     test_labels if len(test_labels) > 0 else np.zeros(len(test_images)), 
                     final_predictions, num_samples=6)

# ========== 6. СОХРАНЕНИЕ РЕЗУЛЬТАТОВ ==========
# Сохранение модели
autoencoder.save('improved_autoencoder_model.keras')
print("\nМодель сохранена как 'improved_autoencoder_model.keras'")

# Сохранение предсказаний
import json
results = []
for i, (filename, mse, pred) in enumerate(zip(test_filenames, test_mse, final_predictions)):
    results.append({
        'filename': filename,
        'mse': float(mse),
        'prediction': int(pred),
        'label': int(test_labels[i]) if i < len(test_labels) else -1
    })

with open('predictions.json', 'w') as f:
    json.dump(results, f, indent=2)

print("Предсказания сохранены в 'predictions.json'")

# ========== 7. ФИНАЛЬНАЯ СВОДКА ==========
print("\n" + "="*50)
print("ФИНАЛЬНАЯ СВОДКА РЕЗУЛЬТАТОВ")
print("="*50)

print(f"\nДанные:")
print(f"  - Тренировочные изображения: {train_images.shape[0]}")
print(f"  - Тестовые изображения: {test_images.shape[0]}")
print(f"  - Изображения с проливами: {proliv_images.shape[0]}")

print(f"\nМодель:")
print(f"  - Архитектура: Улучшенный автоэнкодер")
print(f"  - Параметры: {autoencoder.count_params():,}")
print(f"  - Loss (MSE) на валидации: {history.history['val_loss'][-1]:.6f}")

print(f"\nКачество классификации:")
print(f"  - Используемый порог: {final_threshold:.6f}")
print(f"  - Среднее MSE на тесте: {np.mean(test_mse):.6f}")

if len(test_labels) > 0 and len(np.unique(test_labels)) > 1:
    cm = confusion_matrix(test_labels, final_predictions)
    if cm.size == 4:
        tn, fp, fn, tp = cm.ravel()
        tpr = tp / (tp + fn) if (tp + fn) > 0 else 0
        tnr = tn / (tn + fp) if (tn + fp) > 0 else 0
        accuracy = (tp + tn) / (tp + tn + fp + fn)
        
        print(f"  - TPR (Recall): {tpr:.4f}")
        print(f"  - TNR: {tnr:.4f}")
        print(f"  - Accuracy: {accuracy:.4f}")
        print(f"  - FP: {fp}, FN: {fn}")
        
        # Проверяем выполнение требований
        if tpr >= 0.95 and tnr >= 0.95:
            print(f"\n ТРЕБОВАНИЯ ВЫПОЛНЕНЫ: TPR={tpr:.4f} >= 0.95, TNR={tnr:.4f} >= 0.95")
        else:
            print(f"\n ТРЕБОВАНИЯ НЕ ВЫПОЛНЕНЫ:")
            if tpr < 0.95:
                print(f"   - TPR={tpr:.4f} < 0.95")
            if tnr < 0.95:
                print(f"   - TNR={tnr:.4f} < 0.95")

print(f"\nРаспределение предсказаний:")
unique, counts = np.unique(final_predictions, return_counts=True)
for pred, count in zip(unique, counts):
    label = "Пролив" if pred == 1 else "Нет пролива"
    print(f"  - {label}: {count} ({count/len(final_predictions)*100:.1f}%)")

print("\n" + "="*50)
# ========== 8. ТЕСТИРОВАНИЕ НА ВСЕХ 50 ИЗОБРАЖЕНИЯХ ==========
print("\n" + "="*60)
print("ПРЕДСКАЗАНИЯ ДЛЯ ВСЕХ 50 ТЕСТОВЫХ ИЗОБРАЖЕНИЙ")
print("="*60)

# Загружаем все 50 тестовых изображений заново
test_all_images, test_all_filenames = load_images_from_folder('dataset/test/imgs')
print(f"Загружено всех тестовых изображений: {len(test_all_images)}")

# Вычисляем MSE для всех изображений
test_all_mse = calculate_mse(autoencoder, test_all_images)
test_all_predictions = (test_all_mse > final_threshold).astype(int)

print(f"\nРаспределение предсказаний для всех 50 изображений:")
unique, counts = np.unique(test_all_predictions, return_counts=True)
for pred, count in zip(unique, counts):
    label = "Пролив" if pred == 1 else "Нет пролива"
    print(f"  - {label}: {count} ({count/len(test_all_predictions)*100:.1f}%)")

print(f"\nСтатистика MSE для всех 50 изображений:")
print(f"  - Среднее: {np.mean(test_all_mse):.6f}")
print(f"  - Минимум: {np.min(test_all_mse):.6f}")
print(f"  - Максимум: {np.max(test_all_mse):.6f}")
print(f"  - Стандартное отклонение: {np.std(test_all_mse):.6f}")

# Сохраняем предсказания для всех изображений
all_results = []
for i, (filename, mse, pred) in enumerate(zip(test_all_filenames, test_all_mse, test_all_predictions)):
    all_results.append({
        'filename': filename,
        'mse': float(mse),
        'prediction': int(pred),
        'threshold': float(final_threshold),
        'is_anomaly': bool(pred)
    })

with open('all_predictions.json', 'w') as f:
    json.dump(all_results, f, indent=2)

print(f"\nПредсказания для всех 50 изображений сохранены в 'all_predictions.json'")

# Визуализируем примеры из всех изображений
print("\nВизуализация примеров из всех 50 тестовых изображений:")

# Создаем фиктивные метки для визуализации
dummy_labels = np.full(len(test_all_images), -1)  # -1 означает "неизвестно"

# Обновляем функцию визуализации для работы с неизвестными метками
def visualize_all_results(images, mse_values, predictions, num_samples=8):
    if len(images) == 0:
        print("Нет изображений для визуализации")
        return
        
    if num_samples > len(images):
        num_samples = len(images)
    
    # Выбираем примеры с наибольшим и наименьшим MSE
    sorted_indices = np.argsort(mse_values)
    
    # Берем первые (наименьший MSE) и последние (наибольший MSE)
    low_mse_indices = sorted_indices[:num_samples//2]
    high_mse_indices = sorted_indices[-num_samples//2:]
    indices = np.concatenate([low_mse_indices, high_mse_indices])
    
    if len(indices) == 1:
        fig, axes = plt.subplots(1, 3, figsize=(15, 4))
        axes = [axes]
    else:
        fig, axes = plt.subplots(len(indices), 3, figsize=(15, 4*len(indices)))
    
    for i, idx in enumerate(indices):
        if len(indices) == 1:
            ax0, ax1, ax2 = axes
        else:
            ax0, ax1, ax2 = axes[i]
        
        # Оригинальное изображение
        ax0.imshow(images[idx])
        ax0.set_title(f'Оригинал: {test_all_filenames[idx]}\nMSE: {mse_values[idx]:.6f}')
        ax0.axis('off')
        
        # Реконструкция
        reconstruction = autoencoder.predict(images[idx:idx+1], verbose=0)
        ax1.imshow(reconstruction[0])
        ax1.set_title(f'Реконструкция')
        ax1.axis('off')
        
        # Разность
        diff = np.abs(images[idx] - reconstruction[0])
        ax2.imshow(diff)
        pred_label = "ПРОЛИВ" if predictions[idx] == 1 else "Нет пролива"
        color = "red" if predictions[idx] == 1 else "green"
        ax2.set_title(f'Разность\nПредсказание: {pred_label}', color=color)
        ax2.axis('off')
    
    plt.tight_layout()
    plt.show()

# Визуализируем 8 примеров
visualize_all_results(test_all_images, test_all_mse, test_all_predictions, num_samples=8)

```
