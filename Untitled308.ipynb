```python
# ============================================
# МНОЖЕСТВЕННАЯ КЛАССИФИКАЦИЯ НА CIFAR-10 С RESNET
# ============================================

import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader
import torchvision.transforms as transforms
import torchvision.datasets as datasets
import torchvision.models as models
import numpy as np
import matplotlib.pyplot as plt
from tqdm.auto import tqdm
from pathlib import Path
import random
import pandas as pd
from sklearn.metrics import precision_score, recall_score, f1_score, accuracy_score, confusion_matrix, classification_report
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

# Устанавливаем seed
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed(SEED)
    torch.backends.cudnn.deterministic = True

# ============================================
# 1. КОНФИГУРАЦИЯ
# ============================================

class Config:
    # Датасет
    dataset_name = "CIFAR-10"
    num_classes = 10
    
    # Архитектура
    model_name = "ResNet-18"
    pretrained = False  # Для CIFAR-10 лучше с нуля
    
    # Обучение
    batch_size = 64
    learning_rate = 0.001
    num_epochs = 30
    weight_decay = 1e-4
    
    # Пути
    save_dir = Path("./cifar10_classification_results")
    model_save_path = save_dir / "best_model.pth"
    plots_dir = save_dir / "plots"
    
    # Классы CIFAR-10
    class_names = ['airplane', 'automobile', 'bird', 'cat', 'deer', 
                   'dog', 'frog', 'horse', 'ship', 'truck']

config = Config()
config.save_dir.mkdir(exist_ok=True)
config.plots_dir.mkdir(exist_ok=True)

# ============================================
# 2. ЗАГРУЗКА И ПОДГОТОВКА ДАННЫХ
# ============================================

print(f"Загрузка датасета {config.dataset_name}...")

# Трансформации для данных
train_transform = transforms.Compose([
    transforms.RandomCrop(32, padding=4),
    transforms.RandomHorizontalFlip(),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.4914, 0.4822, 0.4465], 
                        std=[0.2023, 0.1994, 0.2010]),
])

test_transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.4914, 0.4822, 0.4465], 
                        std=[0.2023, 0.1994, 0.2010]),
])

# Загрузка датасета
train_dataset = datasets.CIFAR10(
    root='./data', 
    train=True, 
    download=True, 
    transform=train_transform
)

test_dataset = datasets.CIFAR10(
    root='./data', 
    train=False, 
    download=True, 
    transform=test_transform
)

# Разделение на train и validation
train_size = int(0.8 * len(train_dataset))
val_size = len(train_dataset) - train_size
train_dataset, val_dataset = torch.utils.data.random_split(
    train_dataset, [train_size, val_size]
)

# DataLoader
train_loader = DataLoader(
    train_dataset, 
    batch_size=config.batch_size, 
    shuffle=True, 
    num_workers=2
)

val_loader = DataLoader(
    val_dataset, 
    batch_size=config.batch_size, 
    shuffle=False, 
    num_workers=2
)

test_loader = DataLoader(
    test_dataset, 
    batch_size=config.batch_size, 
    shuffle=False, 
    num_workers=2
)

print(f"Данные загружены:")
print(f"  - Train: {len(train_dataset)} изображений")
print(f"  - Val: {len(val_dataset)} изображений")
print(f"  - Test: {len(test_dataset)} изображений")
print(f"  - Классы: {config.num_classes}")
print(f"  - Имена классов: {config.class_names}")

# ============================================
# 3. МОДЕЛЬ RESNET
# ============================================

def create_resnet_model(num_classes=10, pretrained=False):
    """Создание модели ResNet-18"""
    if pretrained:
        model = models.resnet18(weights=models.ResNet18_Weights.IMAGENET1K_V1)
        # Замораживаем первые слои
        for param in model.parameters():
            param.requires_grad = False
    else:
        model = models.resnet18(weights=None)
    
    # Заменяем последний слой для нашего числа классов
    num_features = model.fc.in_features
    model.fc = nn.Linear(num_features, num_classes)
    
    return model

model = create_resnet_model(num_classes=config.num_classes, pretrained=config.pretrained)

# Перемещаем модель на GPU если доступно
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
model = model.to(device)

print(f"\nМодель: {config.model_name}")
print(f"Устройство: {device}")
print(f"Параметров: {sum(p.numel() for p in model.parameters()):,}")
print(f"Обучаемых параметров: {sum(p.numel() for p in model.parameters() if p.requires_grad):,}")

# ============================================
# 4. ФУНКЦИИ ОБУЧЕНИЯ И ВАЛИДАЦИИ
# ============================================

def train_epoch(model, loader, criterion, optimizer, device):
    """Обучение на одной эпохе"""
    model.train()
    running_loss = 0.0
    correct = 0
    total = 0
    
    pbar = tqdm(loader, desc='Training')
    for inputs, targets in pbar:
        inputs, targets = inputs.to(device), targets.to(device)
        
        optimizer.zero_grad()
        outputs = model(inputs)
        loss = criterion(outputs, targets)
        loss.backward()
        optimizer.step()
        
        running_loss += loss.item()
        _, predicted = outputs.max(1)
        total += targets.size(0)
        correct += predicted.eq(targets).sum().item()
        
        # Обновляем прогресс бар
        pbar.set_postfix({
            'loss': f'{loss.item():.4f}',
            'acc': f'{100.*correct/total:.2f}%'
        })
    
    epoch_loss = running_loss / len(loader)
    epoch_acc = 100. * correct / total
    
    return epoch_loss, epoch_acc

def validate(model, loader, criterion, device):
    """Валидация"""
    model.eval()
    running_loss = 0.0
    correct = 0
    total = 0
    
    with torch.no_grad():
        for inputs, targets in loader:
            inputs, targets = inputs.to(device), targets.to(device)
            outputs = model(inputs)
            loss = criterion(outputs, targets)
            
            running_loss += loss.item()
            _, predicted = outputs.max(1)
            total += targets.size(0)
            correct += predicted.eq(targets).sum().item()
    
    epoch_loss = running_loss / len(loader)
    epoch_acc = 100. * correct / total
    
    return epoch_loss, epoch_acc

def test_model(model, loader, device, class_names):
    """Тестирование модели с расчетом метрик"""
    model.eval()
    all_preds = []
    all_targets = []
    
    with torch.no_grad():
        for inputs, targets in tqdm(loader, desc='Testing'):
            inputs, targets = inputs.to(device), targets.to(device)
            outputs = model(inputs)
            _, predicted = outputs.max(1)
            
            all_preds.extend(predicted.cpu().numpy())
            all_targets.extend(targets.cpu().numpy())
    
    # Общие метрики
    accuracy = accuracy_score(all_targets, all_preds)
    precision = precision_score(all_targets, all_preds, average='weighted', zero_division=0)
    recall = recall_score(all_targets, all_preds, average='weighted', zero_division=0)
    f1 = f1_score(all_targets, all_preds, average='weighted', zero_division=0)
    
    # Метрики по каждому классу
    class_precision = precision_score(all_targets, all_preds, average=None, zero_division=0)
    class_recall = recall_score(all_targets, all_preds, average=None, zero_division=0)
    class_f1 = f1_score(all_targets, all_preds, average=None, zero_division=0)
    
    # Матрица ошибок
    cm = confusion_matrix(all_targets, all_preds)
    
    # Отчет классификации
    report = classification_report(
        all_targets, 
        all_preds, 
        target_names=class_names,
        digits=4
    )
    
    metrics = {
        'accuracy': accuracy,
        'precision': precision,
        'recall': recall,
        'f1': f1,
        'class_precision': class_precision,
        'class_recall': class_recall,
        'class_f1': class_f1,
        'confusion_matrix': cm,
        'classification_report': report,
        'predictions': all_preds,
        'targets': all_targets
    }
    
    return metrics

# ============================================
# 5. ОБУЧЕНИЕ МОДЕЛИ
# ============================================

def train_model(model, train_loader, val_loader, config, device):
    """Полный цикл обучения"""
    print(f"\n{'='*70}")
    print(f"НАЧАЛО ОБУЧЕНИЯ")
    print(f"Архитектура: {config.model_name}")
    print(f"Датасет: {config.dataset_name}")
    print(f"Эпохи: {config.num_epochs}")
    print(f"Batch size: {config.batch_size}")
    print(f"Learning rate: {config.learning_rate}")
    print(f"{'='*70}")
    
    # Loss функция и оптимизатор
    criterion = nn.CrossEntropyLoss()
    optimizer = optim.Adam(
        model.parameters(), 
        lr=config.learning_rate, 
        weight_decay=config.weight_decay
    )
    
    # Scheduler для learning rate
    scheduler = optim.lr_scheduler.StepLR(optimizer, step_size=10, gamma=0.1)
    
    # История обучения
    history = {
        'train_loss': [],
        'train_acc': [],
        'val_loss': [],
        'val_acc': [],
        'learning_rates': []
    }
    
    best_val_acc = 0.0
    best_model_state = None
    
    for epoch in range(config.num_epochs):
        print(f"\nЭпоха [{epoch+1}/{config.num_epochs}]")
        
        # Обучение
        train_loss, train_acc = train_epoch(
            model, train_loader, criterion, optimizer, device
        )
        
        # Валидация
        val_loss, val_acc = validate(model, val_loader, criterion, device)
        
        # Обновление learning rate
        scheduler.step()
        current_lr = optimizer.param_groups[0]['lr']
        
        # Сохранение истории
        history['train_loss'].append(train_loss)
        history['train_acc'].append(train_acc)
        history['val_loss'].append(val_loss)
        history['val_acc'].append(val_acc)
        history['learning_rates'].append(current_lr)
        
        print(f"  Train Loss: {train_loss:.4f}, Train Acc: {train_acc:.2f}%")
        print(f"  Val Loss: {val_loss:.4f}, Val Acc: {val_acc:.2f}%")
        print(f"  Learning Rate: {current_lr:.6f}")
        
        # Сохраняем лучшую модель
        if val_acc > best_val_acc:
            best_val_acc = val_acc
            best_model_state = model.state_dict().copy()
            torch.save({
                'epoch': epoch,
                'model_state_dict': best_model_state,
                'optimizer_state_dict': optimizer.state_dict(),
                'val_acc': best_val_acc,
                'config': config.__dict__
            }, config.model_save_path)
            print(f"  ✓ Сохранена лучшая модель (Val Acc: {val_acc:.2f}%)")
    
    # Загружаем лучшую модель
    if best_model_state:
        model.load_state_dict(best_model_state)
    
    print(f"\nОбучение завершено!")
    print(f"Лучшая точность на валидации: {best_val_acc:.2f}%")
    
    return model, history, best_val_acc

# ============================================
# 6. ВИЗУАЛИЗАЦИЯ РЕЗУЛЬТАТОВ
# ============================================

def plot_training_history(history, save_path):
    """Визуализация истории обучения"""
    fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2, figsize=(14, 10))
    
    epochs = range(1, len(history['train_loss']) + 1)
    
    # Loss
    ax1.plot(epochs, history['train_loss'], 'b-', label='Train', linewidth=2)
    ax1.plot(epochs, history['val_loss'], 'r-', label='Val', linewidth=2)
    ax1.set_xlabel('Эпоха')
    ax1.set_ylabel('Loss')
    ax1.set_title('Loss во время обучения')
    ax1.legend()
    ax1.grid(True, alpha=0.3)
    
    # Accuracy
    ax2.plot(epochs, history['train_acc'], 'b-', label='Train', linewidth=2)
    ax2.plot(epochs, history['val_acc'], 'r-', label='Val', linewidth=2)
    ax2.set_xlabel('Эпоха')
    ax2.set_ylabel('Точность (%)')
    ax2.set_title('Точность во время обучения')
    ax2.legend()
    ax2.grid(True, alpha=0.3)
    
    # Learning Rate
    ax3.plot(epochs, history['learning_rates'], 'g-', linewidth=2)
    ax3.set_xlabel('Эпоха')
    ax3.set_ylabel('Learning Rate')
    ax3.set_title('Learning Rate')
    ax3.grid(True, alpha=0.3)
    
    # Train vs Val Accuracy
    ax4.scatter(history['train_acc'], history['val_acc'], alpha=0.6)
    ax4.plot([0, 100], [0, 100], 'r--', alpha=0.3)
    ax4.set_xlabel('Train Accuracy (%)')
    ax4.set_ylabel('Val Accuracy (%)')
    ax4.set_title('Сравнение Train и Val Accuracy')
    ax4.grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.savefig(save_path, dpi=150, bbox_inches='tight')
    plt.show()

def plot_confusion_matrix(cm, class_names, save_path):
    """Визуализация матрицы ошибок"""
    plt.figure(figsize=(12, 10))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                xticklabels=class_names, 
                yticklabels=class_names)
    plt.title('Матрица ошибок (Confusion Matrix)')
    plt.ylabel('Истинные метки')
    plt.xlabel('Предсказанные метки')
    plt.tight_layout()
    plt.savefig(save_path, dpi=150, bbox_inches='tight')
    plt.show()

def plot_class_metrics(class_metrics, class_names, save_path):
    """Визуализация метрик по классам"""
    fig, axes = plt.subplots(1, 3, figsize=(18, 6))
    
    x = np.arange(len(class_names))
    width = 0.6
    
    # Precision по классам
    axes[0].bar(x, class_metrics['precision'], width, color='skyblue', edgecolor='black')
    axes[0].set_xlabel('Классы')
    axes[0].set_ylabel('Precision')
    axes[0].set_title('Precision по классам')
    axes[0].set_xticks(x)
    axes[0].set_xticklabels(class_names, rotation=45, ha='right')
    axes[0].grid(True, alpha=0.3, axis='y')
    
    # Recall по классам
    axes[1].bar(x, class_metrics['recall'], width, color='lightgreen', edgecolor='black')
    axes[1].set_xlabel('Классы')
    axes[1].set_ylabel('Recall')
    axes[1].set_title('Recall по классам')
    axes[1].set_xticks(x)
    axes[1].set_xticklabels(class_names, rotation=45, ha='right')
    axes[1].grid(True, alpha=0.3, axis='y')
    
    # F1-Score по классам
    axes[2].bar(x, class_metrics['f1'], width, color='salmon', edgecolor='black')
    axes[2].set_xlabel('Классы')
    axes[2].set_ylabel('F1-Score')
    axes[2].set_title('F1-Score по классам')
    axes[2].set_xticks(x)
    axes[2].set_xticklabels(class_names, rotation=45, ha='right')
    axes[2].grid(True, alpha=0.3, axis='y')
    
    plt.tight_layout()
    plt.savefig(save_path, dpi=150, bbox_inches='tight')
    plt.show()

# ============================================
# 7. СОХРАНЕНИЕ РЕЗУЛЬТАТОВ
# ============================================

def save_results(metrics, history, config, test_time):
    """Сохранение всех результатов"""
    
    # Сохраняем метрики в CSV
    results_df = pd.DataFrame({
        'Metric': ['Accuracy', 'Precision', 'Recall', 'F1-Score'],
        'Value': [
            metrics['accuracy'],
            metrics['precision'],
            metrics['recall'],
            metrics['f1']
        ]
    })
    
    # Сохраняем метрики по классам
    class_metrics_df = pd.DataFrame({
        'Class': config.class_names,
        'Precision': metrics['class_precision'],
        'Recall': metrics['class_recall'],
        'F1-Score': metrics['class_f1']
    })
    
    # Сохраняем историю обучения
    history_df = pd.DataFrame({
        'Epoch': range(1, len(history['train_loss']) + 1),
        'Train_Loss': history['train_loss'],
        'Train_Acc': history['train_acc'],
        'Val_Loss': history['val_loss'],
        'Val_Acc': history['val_acc'],
        'Learning_Rate': history['learning_rates']
    })
    
    # Сохраняем все в файлы
    results_df.to_csv(config.save_dir / 'results.csv', index=False)
    class_metrics_df.to_csv(config.save_dir / 'class_metrics.csv', index=False)
    history_df.to_csv(config.save_dir / 'training_history.csv', index=False)
    
    # Сохраняем отчет классификации
    with open(config.save_dir / 'classification_report.txt', 'w') as f:
        f.write("=" * 70 + "\n")
        f.write(f"РЕЗУЛЬТАТЫ КЛАССИФИКАЦИИ\n")
        f.write(f"Датасет: {config.dataset_name}\n")
        f.write(f"Модель: {config.model_name}\n")
        f.write(f"Время тестирования: {test_time:.2f} секунд\n")
        f.write("=" * 70 + "\n\n")
        f.write("ОБЩИЕ МЕТРИКИ:\n")
        f.write(f"  Accuracy:  {metrics['accuracy']:.4f}\n")
        f.write(f"  Precision: {metrics['precision']:.4f}\n")
        f.write(f"  Recall:    {metrics['recall']:.4f}\n")
        f.write(f"  F1-Score:  {metrics['f1']:.4f}\n\n")
        f.write("ДЕТАЛЬНЫЙ ОТЧЕТ:\n")
        f.write(metrics['classification_report'])
    
    print(f"\nРезультаты сохранены в {config.save_dir}/")

# ============================================
# 8. ОСНОВНОЙ БЛОК ВЫПОЛНЕНИЯ
# ============================================

def main():
    """Основная функция"""
    print("=" * 70)
    print("МНОЖЕСТВЕННАЯ КЛАССИФИКАЦИЯ CIFAR-10 С RESNET-18")
    print("=" * 70)
    
    # Обучение модели
    trained_model, history, best_val_acc = train_model(
        model, train_loader, val_loader, config, device
    )
    
    # Визуализация истории обучения
    print("\nВизуализация истории обучения...")
    plot_training_history(history, config.plots_dir / 'training_history.png')
    
    # Тестирование на тестовом наборе
    print("\nТестирование на тестовом наборе...")
    import time
    start_time = time.time()
    test_metrics = test_model(trained_model, test_loader, device, config.class_names)
    test_time = time.time() - start_time
    
    # Вывод результатов
    print("\n" + "=" * 70)
    print("РЕЗУЛЬТАТЫ НА ТЕСТОВОМ НАБОРЕ")
    print("=" * 70)
    print(f"Общие метрики:")
    print(f"  Accuracy:  {test_metrics['accuracy']:.4f}")
    print(f"  Precision: {test_metrics['precision']:.4f}")
    print(f"  Recall:    {test_metrics['recall']:.4f}")
    print(f"  F1-Score:  {test_metrics['f1']:.4f}")
    print(f"\nВремя тестирования: {test_time:.2f} секунд")
    print(f"Количество изображений: {len(test_loader.dataset)}")
    
    # Вывод метрик по классам
    print("\nМетрики по классам:")
    print("-" * 70)
    print(f"{'Класс':<12} {'Precision':<10} {'Recall':<10} {'F1-Score':<10}")
    print("-" * 70)
    for i, class_name in enumerate(config.class_names):
        print(f"{class_name:<12} "
              f"{test_metrics['class_precision'][i]:<10.4f} "
              f"{test_metrics['class_recall'][i]:<10.4f} "
              f"{test_metrics['class_f1'][i]:<10.4f}")
    
    # Визуализация матрицы ошибок
    print("\nВизуализация матрицы ошибок...")
    plot_confusion_matrix(
        test_metrics['confusion_matrix'], 
        config.class_names,
        config.plots_dir / 'confusion_matrix.png'
    )
    
    # Визуализация метрик по классам
    print("Визуализация метрик по классам...")
    class_metrics = {
        'precision': test_metrics['class_precision'],
        'recall': test_metrics['class_recall'],
        'f1': test_metrics['class_f1']
    }
    plot_class_metrics(
        class_metrics,
        config.class_names,
        config.plots_dir / 'class_metrics.png'
    )
    
    # Сохранение результатов
    print("Сохранение результатов...")
    save_results(test_metrics, history, config, test_time)
    
    # Вывод отчета классификации
    print("\n" + "=" * 70)
    print("ПОЛНЫЙ ОТЧЕТ КЛАССИФИКАЦИИ")
    print("=" * 70)
    print(test_metrics['classification_report'])
    
    print("\n" + "=" * 70)
    print("ЭКСПЕРИМЕНТ УСПЕШНО ЗАВЕРШЕН!")
    print(f"Все результаты сохранены в папке: {config.save_dir}")
    print("=" * 70)
    
    return trained_model, test_metrics, history

# ============================================
# 9. ЗАПУСК ПРОГРАММЫ
# ============================================

if __name__ == "__main__":
    trained_model, test_metrics, history = main()
```
