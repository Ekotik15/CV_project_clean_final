```python
# ==================== ИМПОРТ БИБЛИОТЕК ====================
import torch
import torch.nn as nn
import torch.optim as optim
import torchvision
import torchvision.transforms as transforms
from torch.utils.data import DataLoader, random_split, Subset
import os
import matplotlib.pyplot as plt
import numpy as np
from sklearn.metrics import classification_report
from tqdm import tqdm
import random
from IPython.display import clear_output
import warnings
warnings.filterwarnings('ignore')

# Устанавливаем seed для воспроизводимости
def set_seed(seed=42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False

set_seed(42)

# ==================== ОПРЕДЕЛЕНИЕ АРХИТЕКТУРЫ ====================
class BasicBlock(nn.Module):
    def __init__(self, in_planes, planes, stride=1):
        super(BasicBlock, self).__init__()
        self.conv1 = nn.Conv2d(in_planes, planes, kernel_size=3, stride=stride, padding=1, bias=False)
        self.bn1 = nn.BatchNorm2d(planes)
        self.conv2 = nn.Conv2d(planes, planes, kernel_size=3, stride=1, padding=1, bias=False)
        self.bn2 = nn.BatchNorm2d(planes)
        self.shortcut = nn.Sequential()
        if stride != 1 or in_planes != planes:
            self.shortcut = nn.Sequential(
                nn.Conv2d(in_planes, planes, kernel_size=1, stride=stride, bias=False),
                nn.BatchNorm2d(planes)
            )

    def forward(self, x):
        out = torch.relu(self.bn1(self.conv1(x)))
        out = self.bn2(self.conv2(out))
        out += self.shortcut(x)
        out = torch.relu(out)
        return out

class ResNet(nn.Module):
    def __init__(self, block, num_blocks, num_classes=10):
        super(ResNet, self).__init__()
        self.in_planes = 64
        self.conv1 = nn.Conv2d(3, 64, kernel_size=3, stride=1, padding=1, bias=False)
        self.bn1 = nn.BatchNorm2d(64)
        self.layer1 = self._make_layer(block, 64, num_blocks[0], stride=1)
        self.layer2 = self._make_layer(block, 128, num_blocks[1], stride=2)
        self.layer3 = self._make_layer(block, 256, num_blocks[2], stride=2)
        self.layer4 = self._make_layer(block, 512, num_blocks[3], stride=2)
        self.linear = nn.Linear(512, num_classes)

    def _make_layer(self, block, planes, num_blocks, stride):
        strides = [stride] + [1] * (num_blocks - 1)
        layers = []
        for stride in strides:
            layers.append(block(self.in_planes, planes, stride))
            self.in_planes = planes
        return nn.Sequential(*layers)

    def forward(self, x):
        out = torch.relu(self.bn1(self.conv1(x)))
        out = self.layer1(out)
        out = self.layer2(out)
        out = self.layer3(out)
        out = self.layer4(out)
        out = torch.nn.functional.avg_pool2d(out, 4)
        out = out.view(out.size(0), -1)
        out = self.linear(out)
        return out

def ResNet18(num_classes=10):
    return ResNet(BasicBlock, [2, 2, 2, 2], num_classes=num_classes)

# ==================== SSL КОМПОНЕНТЫ ====================
class RotationTransform:
    """Трансформация для SSL задачи предсказания поворота"""
    def __init__(self):
        self.angles = [0, 90, 180, 270]
    
    def __call__(self, x):
        angle = random.choice(self.angles)
        return transforms.functional.rotate(x, angle), self.angles.index(angle)

class SSLModel(nn.Module):
    """Self-supervised модель для предсказания поворота"""
    def __init__(self, base_model, num_classes=4):
        super(SSLModel, self).__init__()
        self.feature_extractor = base_model
        self.feature_extractor.linear = nn.Linear(512, num_classes)
        
    def forward(self, x):
        return self.feature_extractor(x)

class ResNetSSL(nn.Module):
    """ResNet с возможностью заморозки feature extractor"""
    def __init__(self, base_model, freeze_features=True):
        super(ResNetSSL, self).__init__()
        self.features = nn.Sequential(
            base_model.conv1,
            base_model.bn1,
            nn.ReLU(),
            base_model.layer1,
            base_model.layer2,
            base_model.layer3,
            base_model.layer4,
            nn.AdaptiveAvgPool2d((1, 1))
        )
        
        if freeze_features:
            for param in self.features.parameters():
                param.requires_grad = False
        
        self.classifier = nn.Linear(512, 10)
        
    def forward(self, x):
        x = self.features(x)
        x = torch.flatten(x, 1)
        x = self.classifier(x)
        return x

# ==================== ВСПОМОГАТЕЛЬНЫЕ ФУНКЦИИ ====================
def get_subset_with_percentage(dataset, percentage=100, num_classes=10):
    """Получает подмножество датасета с заданным процентом данных"""
    if percentage == 100:
        return dataset
    
    indices_by_class = {i: [] for i in range(num_classes)}
    for idx, (_, label) in enumerate(dataset):
        indices_by_class[label].append(idx)
    
    selected_indices = []
    for class_idx in indices_by_class:
        class_indices = indices_by_class[class_idx]
        n_select = max(1, int(len(class_indices) * percentage / 100))
        selected_indices.extend(random.sample(class_indices, n_select))
    
    return Subset(dataset, selected_indices)

def find_plateau_epoch(accuracies, window=5, threshold=0.1):
    """Находит эпоху, когда модель достигает плато"""
    if len(accuracies) < window:
        return len(accuracies)
    
    for i in range(len(accuracies) - window):
        window_vals = accuracies[i:i+window]
        if max(window_vals) - min(window_vals) < threshold:
            return i + window//2
    
    return len(accuracies)

# ==================== ФУНКЦИИ ОБУЧЕНИЯ ====================
def train_ssl_model(model, train_loader, device, epochs=30, lr=0.001):
    """Предобучение feature extractor с помощью SSL"""
    print("\n" + "="*60)
    print("SELF-SUPERVISED PRE-TRAINING (Rotation Prediction)")
    print("="*60)
    
    criterion = nn.CrossEntropyLoss()
    optimizer = optim.Adam(model.parameters(), lr=lr)
    scheduler = optim.lr_scheduler.StepLR(optimizer, step_size=10, gamma=0.1)
    
    model.train()
    losses, accuracies = [], []
    
    for epoch in range(epochs):
        running_loss, correct, total = 0.0, 0, 0
        
        pbar = tqdm(train_loader, desc=f'SSL Epoch {epoch+1}/{epochs}')
        for batch_idx, (inputs, targets) in enumerate(pbar):
            inputs, rotation_labels = inputs
            inputs, rotation_labels = inputs.to(device), rotation_labels.to(device)
            
            optimizer.zero_grad()
            outputs = model(inputs)
            loss = criterion(outputs, rotation_labels)
            loss.backward()
            optimizer.step()
            
            running_loss += loss.item()
            _, predicted = outputs.max(1)
            total += rotation_labels.size(0)
            correct += predicted.eq(rotation_labels).sum().item()
            
            pbar.set_postfix({
                'Loss': f'{running_loss/(batch_idx+1):.3f}',
                'Acc': f'{100.*correct/total:.2f}%'
            })
        
        scheduler.step()
        epoch_loss = running_loss / len(train_loader)
        epoch_acc = 100. * correct / total
        losses.append(epoch_loss)
        accuracies.append(epoch_acc)
        
        print(f'SSL Epoch {epoch+1}: Loss = {epoch_loss:.4f}, Accuracy = {epoch_acc:.2f}%')
    
    # Визуализация обучения SSL
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))
    ax1.plot(losses); ax1.set_title('SSL Training Loss'); ax1.set_xlabel('Epoch'); ax1.set_ylabel('Loss'); ax1.grid(True)
    ax2.plot(accuracies); ax2.set_title('SSL Training Accuracy'); ax2.set_xlabel('Epoch'); ax2.set_ylabel('Accuracy (%)'); ax2.grid(True)
    plt.tight_layout(); plt.show()
    
    print("Self-supervised pre-training completed!")
    return model

def train_supervised(model, train_loader, val_loader, device, epochs=50, lr=0.001, use_ssl=False):
    """Supervised обучение модели"""
    criterion = nn.CrossEntropyLoss()
    
    if use_ssl:
        optimizer = optim.Adam(model.classifier.parameters(), lr=lr)
    else:
        optimizer = optim.Adam(model.parameters(), lr=lr, weight_decay=1e-4)
    
    scheduler = optim.lr_scheduler.StepLR(optimizer, step_size=20, gamma=0.1)
    
    history = {'train_losses': [], 'train_accs': [], 'val_losses': [], 'val_accs': []}
    best_val_acc = 0.0
    
    for epoch in range(epochs):
        # Обучение
        model.train()
        running_loss, correct, total = 0.0, 0, 0
        
        pbar = tqdm(train_loader, desc=f'Epoch {epoch+1}/{epochs}')
        for batch_idx, (inputs, targets) in enumerate(pbar):
            inputs, targets = inputs.to(device), targets.to(device)
            
            optimizer.zero_grad()
            outputs = model(inputs)
            loss = criterion(outputs, targets)
            loss.backward()
            optimizer.step()
            
            running_loss += loss.item()
            _, predicted = outputs.max(1)
            total += targets.size(0)
            correct += predicted.eq(targets).sum().item()
            
            pbar.set_postfix({
                'Loss': f'{running_loss/(batch_idx+1):.3f}',
                'Acc': f'{100.*correct/total:.2f}%'
            })
        
        train_loss = running_loss / len(train_loader)
        train_acc = 100. * correct / total
        
        # Валидация
        model.eval()
        running_loss, correct, total = 0.0, 0, 0
        
        with torch.no_grad():
            for inputs, targets in val_loader:
                inputs, targets = inputs.to(device), targets.to(device)
                outputs = model(inputs)
                loss = criterion(outputs, targets)
                running_loss += loss.item()
                _, predicted = outputs.max(1)
                total += targets.size(0)
                correct += predicted.eq(targets).sum().item()
        
        val_loss = running_loss / len(val_loader)
        val_acc = 100. * correct / total
        
        # Сохранение истории
        history['train_losses'].append(train_loss)
        history['train_accs'].append(train_acc)
        history['val_losses'].append(val_loss)
        history['val_accs'].append(val_acc)
        
        scheduler.step()
        
        print(f'Training Loss: {train_loss:.4f}, Accuracy: {train_acc:.2f}%')
        print(f'Validation Loss: {val_loss:.4f}, Accuracy: {val_acc:.2f}%')
        
        if val_acc > best_val_acc:
            best_val_acc = val_acc
    
    return history, best_val_acc

def test_model(model, test_loader, device):
    """Тестирование модели"""
    model.eval()
    correct, total = 0, 0
    all_preds, all_targets = [], []
    
    with torch.no_grad():
        for inputs, targets in test_loader:
            inputs, targets = inputs.to(device), targets.to(device)
            outputs = model(inputs)
            _, predicted = outputs.max(1)
            total += targets.size(0)
            correct += predicted.eq(targets).sum().item()
            all_preds.extend(predicted.cpu().numpy())
            all_targets.extend(targets.cpu().numpy())
    
    test_acc = 100. * correct / total
    return test_acc, all_preds, all_targets

def compare_results(baseline_history, ssl_history, baseline_test_acc, ssl_test_acc, percentage):
    """Сравнивает результаты обучения"""
    print("\n" + "="*60)
    print(f"COMPARISON RESULTS ({percentage}% DATA)")
    print("="*60)
    
    baseline_best_val = max(baseline_history['val_accs'])
    ssl_best_val = max(ssl_history['val_accs'])
    
    print(f"Baseline - Best Val Accuracy: {baseline_best_val:.2f}%, Test Accuracy: {baseline_test_acc:.2f}%")
    print(f"SSL Pre-trained - Best Val Accuracy: {ssl_best_val:.2f}%, Test Accuracy: {ssl_test_acc:.2f}%")
    print(f"Improvement: {ssl_test_acc - baseline_test_acc:.2f}%")
    
    # Определяем плато
    baseline_plateau = find_plateau_epoch(baseline_history['val_accs'])
    ssl_plateau = find_plateau_epoch(ssl_history['val_accs'])
    
    print(f"\nBaseline reached plateau at epoch: {baseline_plateau}")
    print(f"SSL Pre-trained reached plateau at epoch: {ssl_plateau}")
    
    if ssl_plateau < baseline_plateau:
        print("✓ SSL Pre-trained reached plateau FASTER")
    elif ssl_plateau > baseline_plateau:
        print("✗ Baseline reached plateau FASTER")
    else:
        print("Both reached plateau at the same time")
    
    # Визуализация
    fig, axes = plt.subplots(2, 2, figsize=(15, 10))
    
    axes[0, 0].plot(baseline_history['train_losses'], label='Baseline', linewidth=2)
    axes[0, 0].plot(ssl_history['train_losses'], label='SSL Pre-trained', linewidth=2)
    axes[0, 0].set_title(f'Training Loss ({percentage}% data)')
    axes[0, 0].set_xlabel('Epoch'); axes[0, 0].set_ylabel('Loss'); axes[0, 0].legend(); axes[0, 0].grid(True)
    
    axes[0, 1].plot(baseline_history['train_accs'], label='Baseline', linewidth=2)
    axes[0, 1].plot(ssl_history['train_accs'], label='SSL Pre-trained', linewidth=2)
    axes[0, 1].set_title(f'Training Accuracy ({percentage}% data)')
    axes[0, 1].set_xlabel('Epoch'); axes[0, 1].set_ylabel('Accuracy (%)'); axes[0, 1].legend(); axes[0, 1].grid(True)
    
    axes[1, 0].plot(baseline_history['val_losses'], label='Baseline', linewidth=2)
    axes[1, 0].plot(ssl_history['val_losses'], label='SSL Pre-trained', linewidth=2)
    axes[1, 0].set_title(f'Validation Loss ({percentage}% data)')
    axes[1, 0].set_xlabel('Epoch'); axes[1, 0].set_ylabel('Loss'); axes[1, 0].legend(); axes[1, 0].grid(True)
    
    axes[1, 1].plot(baseline_history['val_accs'], label='Baseline', linewidth=2)
    axes[1, 1].plot(ssl_history['val_accs'], label='SSL Pre-trained', linewidth=2)
    axes[1, 1].set_title(f'Validation Accuracy ({percentage}% data)')
    axes[1, 1].set_xlabel('Epoch'); axes[1, 1].set_ylabel('Accuracy (%)'); axes[1, 1].legend(); axes[1, 1].grid(True)
    
    plt.tight_layout(); plt.savefig(f'comparison_{percentage}.png', dpi=150, bbox_inches='tight'); plt.show()
    
    return baseline_best_val, ssl_best_val, baseline_plateau, ssl_plateau

# ==================== ОСНОВНОЙ ЭКСПЕРИМЕНТ ====================
def run_experiment(percentage=100, batch_size=128, epochs=50, ssl_epochs=30):
    """Запускает эксперимент для заданного процента данных"""
    device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    print(f"\n{'='*80}")
    print(f"ЭКСПЕРИМЕНТ: {percentage}% РАЗМЕЧЕННЫХ ДАННЫХ")
    print(f"{'='*80}")
    print(f"Используется устройство: {device}")
    
    # Создаем директории
    os.makedirs('./data', exist_ok=True)
    
    # Трансформации
    transform_train = transforms.Compose([
        transforms.RandomCrop(32, padding=4),
        transforms.RandomHorizontalFlip(),
        transforms.ToTensor(),
        transforms.Normalize((0.4914, 0.4822, 0.4465), (0.2023, 0.1994, 0.2010))
    ])
    
    transform_test = transforms.Compose([
        transforms.ToTensor(),
        transforms.Normalize((0.4914, 0.4822, 0.4465), (0.2023, 0.1994, 0.2010))
    ])
    
    # Загрузка данных
    print("\nЗагрузка датасета CIFAR-10...")
    full_train_dataset = torchvision.datasets.CIFAR10(
        root='./data', train=True, download=True, transform=transform_train
    )
    
    test_dataset = torchvision.datasets.CIFAR10(
        root='./data', train=False, download=True, transform=transform_test
    )
    
    # Класс для SSL датасета
    class RotationDataset(torch.utils.data.Dataset):
        def __init__(self, dataset):
            self.dataset = dataset
            self.rotation_transform = RotationTransform()
            
        def __len__(self):
            return len(self.dataset)
            
        def __getitem__(self, idx):
            img, _ = self.dataset[idx]
            rotated_img, rotation_label = self.rotation_transform(img)
            return (rotated_img, rotation_label), (img, _)
    
    # ==================== SSL ПРЕДОБУЧЕНИЕ ====================
    print("\n" + "="*40)
    print("1. SELF-SUPERVISED PRE-TRAINING")
    print("="*40)
    
    # Создаем модель для SSL
    ssl_base_model = ResNet18(num_classes=4)
    ssl_model = SSLModel(ssl_base_model).to(device)
    
    # Создаем датасет для SSL
    ssl_dataset = RotationDataset(full_train_dataset)
    ssl_loader = DataLoader(ssl_dataset, batch_size=batch_size, shuffle=True, num_workers=0)
    
    # Предобучение на задаче поворота
    ssl_model = train_ssl_model(ssl_model, ssl_loader, device, epochs=ssl_epochs)
    
    # ==================== BASELINE: ОБУЧЕНИЕ С НУЛЯ ====================
    print("\n" + "="*40)
    print("2. BASELINE: ОБУЧЕНИЕ С НУЛЯ")
    print("="*40)
    
    # Подготовка данных
    labeled_dataset = get_subset_with_percentage(full_train_dataset, percentage)
    val_size = int(0.1 * len(labeled_dataset))
    train_size = len(labeled_dataset) - val_size
    train_dataset, val_dataset = random_split(labeled_dataset, [train_size, val_size])
    val_dataset.dataset.transform = transform_test
    
    train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True, num_workers=0)
    val_loader = DataLoader(val_dataset, batch_size=batch_size, shuffle=False, num_workers=0)
    test_loader = DataLoader(test_dataset, batch_size=batch_size, shuffle=False, num_workers=0)
    
    print(f"Обучающих примеров: {len(train_dataset)}")
    print(f"Валидационных примеров: {len(val_dataset)}")
    print(f"Тестовых примеров: {len(test_dataset)}")
    
    # Модель для baseline
    baseline_model = ResNet18(num_classes=10).to(device)
    
    # Обучение baseline
    baseline_history, baseline_best_val = train_supervised(
        baseline_model, train_loader, val_loader, device, epochs=epochs, use_ssl=False
    )
    
    # Тестирование baseline
    baseline_test_acc, baseline_preds, baseline_targets = test_model(baseline_model, test_loader, device)
    print(f"\nBaseline Test Accuracy: {baseline_test_acc:.2f}%")
    
    # ==================== SSL PRE-TRAINED: ОБУЧЕНИЕ С ПРЕДОБУЧЕННЫМ FEATURE EXTRACTOR ====================
    print("\n" + "="*40)
    print("3. SSL PRE-TRAINED: ОБУЧЕНИЕ С ПРЕДОБУЧЕННЫМ FEATURE EXTRACTOR")
    print("="*40)
    
    # Создаем основную модель с замороженным feature extractor
    base_model = ResNet18(num_classes=10)
    ssl_state_dict = ssl_model.state_dict()
    
    # Удаляем веса последнего слоя SSL модели
    for key in list(ssl_state_dict.keys()):
        if 'linear' in key:
            del ssl_state_dict[key]
    
    # Загружаем веса в основную модель
    base_model.load_state_dict(ssl_state_dict, strict=False)
    ssl_pretrained_model = ResNetSSL(base_model, freeze_features=True).to(device)
    
    # Обучение SSL модели
    ssl_history, ssl_best_val = train_supervised(
        ssl_pretrained_model, train_loader, val_loader, device, epochs=epochs, use_ssl=True
    )
    
    # Тестирование SSL модели
    ssl_test_acc, ssl_preds, ssl_targets = test_model(ssl_pretrained_model, test_loader, device)
    print(f"\nSSL Pre-trained Test Accuracy: {ssl_test_acc:.2f}%")
    
    # ==================== СРАВНЕНИЕ РЕЗУЛЬТАТОВ ====================
    print("\n" + "="*40)
    print("4. СРАВНЕНИЕ РЕЗУЛЬТАТОВ")
    print("="*40)
    
    baseline_best_val, ssl_best_val, baseline_plateau, ssl_plateau = compare_results(
        baseline_history, ssl_history, baseline_test_acc, ssl_test_acc, percentage
    )
    
    # Классификационные отчеты
    class_names = ['самолет', 'автомобиль', 'птица', 'кошка', 'олень',
                   'собака', 'лягушка', 'лошадь', 'корабль', 'грузовик']
    
    print("\nBaseline Classification Report:")
    print(classification_report(baseline_targets, baseline_preds, target_names=class_names, digits=3))
    
    print("\nSSL Pre-trained Classification Report:")
    print(classification_report(ssl_targets, ssl_preds, target_names=class_names, digits=3))
    
    return {
        'percentage': percentage,
        'baseline': {
            'history': baseline_history,
            'test_acc': baseline_test_acc,
            'best_val_acc': baseline_best_val,
            'plateau_epoch': baseline_plateau
        },
        'ssl': {
            'history': ssl_history,
            'test_acc': ssl_test_acc,
            'best_val_acc': ssl_best_val,
            'plateau_epoch': ssl_plateau
        }
    }

# ==================== ЗАПУСК ВСЕХ ЭКСПЕРИМЕНТОВ ====================
def run_all_experiments():
    """Запускает все эксперименты для N = 100, 50, 10"""
    percentages = [100, 50, 10]
    results = {}
    
    for percentage in percentages:
        results[percentage] = run_experiment(
            percentage=percentage,
            batch_size=128,
            epochs=30,  # Уменьшим для быстрого выполнения
            ssl_epochs=20
        )
        
        # Сохраняем промежуточные результаты
        torch.save(results, 'experiment_results.pth')
    
    # ==================== ИТОГОВЫЙ ОТЧЕТ ====================
    print("\n" + "="*80)
    print("ИТОГОВЫЙ ОТЧЕТ ПО ВСЕМ ЭКСПЕРИМЕНТАМ")
    print("="*80)
    
    print("\nСРАВНЕНИЕ ТОЧНОСТИ:")
    print("-"*60)
    print(f"{'Процент данных':<15} {'Baseline':<12} {'SSL':<12} {'Улучшение':<12}")
    print("-"*60)
    
    for percentage in percentages:
        baseline_acc = results[percentage]['baseline']['test_acc']
        ssl_acc = results[percentage]['ssl']['test_acc']
        improvement = ssl_acc - baseline_acc
        print(f"{percentage}%:          {baseline_acc:<11.2f}% {ssl_acc:<11.2f}% {improvement:<11.2f}%")
    
    print("\nСРАВНЕНИЕ СКОРОСТИ ДОСТИЖЕНИЯ ПЛАТО:")
    print("-"*60)
    print(f"{'Процент данных':<15} {'Baseline':<12} {'SSL':<12} {'Быстрее':<12}")
    print("-"*60)
    
    for percentage in percentages:
        baseline_plateau = results[percentage]['baseline']['plateau_epoch']
        ssl_plateau = results[percentage]['ssl']['plateau_epoch']
        faster = "SSL" if ssl_plateau < baseline_plateau else "Baseline" if baseline_plateau < ssl_plateau else "Equal"
        print(f"{percentage}%:          {baseline_plateau:<11} {ssl_plateau:<11} {faster:<11}")
    
    # Итоговый график
    fig, axes = plt.subplots(1, 3, figsize=(18, 5))
    
    for idx, percentage in enumerate(percentages):
        baseline_history = results[percentage]['baseline']['history']
        ssl_history = results[percentage]['ssl']['history']
        
        axes[idx].plot(baseline_history['val_accs'], label='Baseline', linewidth=2)
        axes[idx].plot(ssl_history['val_accs'], label='SSL Pre-trained', linewidth=2)
        axes[idx].set_title(f'{percentage}% размеченных данных')
        axes[idx].set_xlabel('Эпоха')
        axes[idx].set_ylabel('Точность на валидации (%)')
        axes[idx].legend()
        axes[idx].grid(True)
        axes[idx].set_ylim([0, 100])
    
    plt.tight_layout()
    plt.savefig('final_results.png', dpi=150, bbox_inches='tight')
    plt.show()
    
    print("\n" + "="*80)
    print("ЭКСПЕРИМЕНТЫ ЗАВЕРШЕНЫ!")
    print("="*80)
    print("\nСозданные файлы:")
    print("- comparison_100.png, comparison_50.png, comparison_10.png: Графики сравнения")
    print("- final_results.png: Итоговый график")
    print("- experiment_results.pth: Сохраненные результаты")
    
    return results

# ==================== ЗАПУСК ====================
if __name__ == '__main__':
    print("Self-Supervised Learning Experiments: CIFAR-10")
    print("Сравнение обучения с предобучением на задаче поворота и без")
    print("\nЗапускаем эксперименты для N = 100%, 50%, 10% размеченных данных...")
    
    # Для ускорения можно уменьшить количество эпох
    results = run_all_experiments()
```
